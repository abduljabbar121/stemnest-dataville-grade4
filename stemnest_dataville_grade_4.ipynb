```python
# task1.py - G4_DV_L06_[Abduljabbar]_Comments
# Grade 4 Dataville - Lesson 07
# District Population Analysis - Professional Comments
# Analyst: [Abduljabbar]  |   Date:[19/10/2026]
"""
DATA SOURCE: Dataville National Census, Q3 2024
SCOPE: Five adminstrative districts
PURPOSE: Compute summary statistics for Dr Osei's Q4 infrastructure
planning report
"""

# === DISTRICT POPULATION ===
# Source: census_Q3_2024.csv, column: resident_population
accra_central   = 2_500_000  # highest-density urban district
kumasi_district = 1_800_000  # second-largest by population
tamale_north    =   950_000  # largest district by land area
sekondi_takoradi =  750_000  # primary coastal industrial district
cape_coast       =  250_000  # historical capital, lower density
district_count   = 5         # total districts in this anlysis

# --- CALCULATIONS ---
# Total: sum for national-level reporting
total_population  =  accra_central + kumasi_district + tamale_north + sekondi_takoradi + cape_coast
# Average: mean for per-district resources allocation model
average_population = total_population / district_count
# Range: max minus min - indicator of population inequality
population_range  = accra_central - cape_coast

# === OUTPUT ===
# Report formatted for Dr Osei's dashboard
print('=' * 42)
print('DATAVILLE POPULATION ANALYSIS')
print('Source: Census Q3 2024')
print('=' * 42)
print('Total:    ', total_population)
print('Average:  ', average_population)
print('Range:    ', population_range)
print('=' * 42)
print('Analysis complete - ready for dashboard.')
```

```python
# Three-District Comparison Tool
# task1.py - G4_DV_L08_[ABDUL-JABBAR]_Integration
# Grade 4 Dataville - Lesson 08
# Three-District Comparison Tool
# Purpose: Accept district data from user, produce comparison report
"""
Collects name, population (string), and data quality score (string) for three Dataville districts. Produces a comparison report.
Note: Population and quality stay as strings (conversion in Unit 2).
"""

print('=' * 48)
print('DATAVILLE THREE-DISTRICT COMPARISON TOOL')
print('=' * 48)
print()

# INPUT - DISTRICT 1
print('--- District 1 ---')
d1_name     = input('    Name:  ')
d1_pop      = input('   Population:     ')
d1_quality  = input('  Data quailty (1-10):  ')
d1_industry = input('  Main industry:   ')
print()

# INPUT - DISTRICT 2
print('--- District 2 ---')
d2_name     = input('   Name:  ')
d2_pop      = input('   Population:     ')
d2_quality  = input('  Data quailty (1-10):  ')
d2_industry = input('  Main industry:   ')
print()

# INPUT - DISTRICT 3
print('--- District 3 ---')
d3_name     = input('   Name:  ')
d3_pop      = input('   Population:     ')
d3_quality  = input('  Data quailty (1-10):  ')
d3_industry = input('  Main industry:   ')
print()

# OUTPUT - Comparison report
sep_heavy = '=' * 48
sep_light = '=' * 48
print()
print(sep_heavy)
print('THREE-DISTRICT COMPARISON REPORT')
print('Prepared by: Junior Analys for Dr Amara Osei')
print(sep_heavy)
print(f'Distrcit:    (d1_name)')
print(f'Population:  (d1_pop)')
print(f'Quality:     (d1_quality)/10')
print(f'Industry:    (d1_industry)')
print(sep_light)
print(f'Distrcit:    (d2_name)')
print(f'Population:  (d2_pop)')
print(f'Quality:     (d2_quality)/10')
print(f'Industry:    (d2_industry)')
print(sep_light)
print(f'Distrcit:    (d3_name)')
print(f'Population:  (d3_pop)')
print(f'Quality:     (d3_quality)/10')
print(f'Industry:    (d3_industry)')
print(sep_heavy)
print('Report complete. Ready for dashboard submission.')
```
