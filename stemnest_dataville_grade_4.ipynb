```python
# task1.py - G4_DV_L06_[Abdul-Jabbar]_Calculations
# District population calculator - variables and arithmetic

# Dataville populations
accra_central    = 2_500_000
kumasi_district  = 1_800_000
tamale_north     =  950_000
sekondi_takoradi =  750_000
cape_coast       =  250_000
district_count   =  5

# Calculations
total_population  = accra_central + kumasi_district + tamale_north + sekondi_takoradi + cape_coast
average_population = total_population / district_count
population_range = accra_central - cape_coast

# Output
print('=' * 42)
print('DATAVILLE POPULATION ANALYSIS')
print('=' * 42)
print('Accra Central:    ',    accra_central)
print('Kumasi:           ',    kumasi_district)
print('Temale North      ',    tamale_north)
print('Sekondi-Takoradi  ',    sekondi_takoradi)
print('Cape Coast:       ',    cape_coast)
print()
print('Total population: ',    total_population)
print('Average:          ',    average_population)
print('Range:            ',    population_range)
print('=' * 42)
```
