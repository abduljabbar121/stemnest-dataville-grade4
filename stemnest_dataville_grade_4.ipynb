```python
# task1.py - G4_DV_L06_[Abduljabbar]_Comments
# Grade 4 Dataville - Lesson 07
# District Population Analysis - Professional Comments
# Analyst: [Abduljabbar]  |   Date:[19/10/2026]
"""
DATA SOURCE: Dataville National Census, Q3 2024
SCOPE: Five adminstrative districts
PURPOSE: Compute summary statistics for Dr Osei's Q4 infrastructure
planning report
"""

# === DISTRICT POPULATION ===
# Source: census_Q3_2024.csv, column: resident_population
accra_central   = 2_500_000  # highest-density urban district
kumasi_district = 1_800_000  # second-largest by population
tamale_north    =   950_000  # largest district by land area
sekondi_takoradi =  750_000  # primary coastal industrial district
cape_coast       =  250_000  # historical capital, lower density
district_count   = 5         # total districts in this anlysis

# --- CALCULATIONS ---
# Total: sum for national-level reporting
total_population  =  accra_central + kumasi_district + tamale_north + sekondi_takoradi + cape_coast
# Average: mean for per-district resources allocation model
average_population = total_population / district_count
# Range: max minus min - indicator of population inequality
population_range  = accra_central - cape_coast

# === OUTPUT ===
# Report formatted for Dr Osei's dashboard
print('=' * 42)
print('DATAVILLE POPULATION ANALYSIS')
print('Source: Census Q3 2024')
print('=' * 42)
print('Total:    ', total_population)
print('Average:  ', average_population)
print('Range:    ', population_range)
print('=' * 42)
print('Analysis complete - ready for dashboard.')
```
